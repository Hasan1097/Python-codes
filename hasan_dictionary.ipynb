```python
# Problem 7: Filter Dictionary by Value Threshold
# Difficulty: Medium

# Problem Statement
# Return only those students whose marks are above a given threshold.
# Input Format
# {'Ali': 88, 'Sara': 45, 'John': 92}
# Threshold = 70
# Output Format
# {'Ali': 88, 'John': 92}

a = {'Ali': 88, 'Sara': 45, 'John': 92}
threshold = 70
b = {}
for name,marks in a.items():
  if marks > threshold:
    b[name] = marks

print(b)
```

```python
# Problem 8: Word Frequency Counter
# Difficulty: Medium

# Problem Statement
# Count how many times each word appears in a sentence (ignore case).
# Input Format
# This is a test. This test is simple.
# Output Format
# {'this': 2, 'is': 2, 'a': 1, 'test': 2, 'simple': 1}

# variable = ("this is a test this test is simple").replace(".","").split()

# new = {}

# for word,count in enumerate(variable):
#   new[word] = count
# print(new)


sentence = "This is a test. This test is simple."

# normalize the input
words = sentence.lower().replace(".", "").split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
        print(frequency[word])
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Problem 9: Reverse Lookup in Dictionary
# Difficulty: Medium

# Problem Statement
# Find all keys that map to the given value.
# Input Format
# {'a': 1, 'b': 2, 'c': 1}
# Target = 1
# Output Format
# ['a', 'c']

variable = {'a': 1, 'b': 2, 'c': 1}
target = 1
new_variable = {}
for word,value in variable.items():
  if value == target:
    new_variable[word] = value

print(new_variable.keys())

```

```python
# Problem 10: Inventory Availability Checker
# Difficulty: Medium

# Problem Statement
# Determine which requested items are available in inventory.
# Input Format
# {'apple':10,'banana':0,'milk':5}
# ['apple','banana','bread']
# Output Format
# Available: ['apple', 'milk']
# Unavailable: ['banana', 'bread']

inventory = {'apple':10,'banana':0,'milk':5}

demand = ['apple','banana','bread']
available = []
unavailable = []

for item in demand:
  print(item)
  if item in inventory and inventory[item] > 0:
    available.append(item)
  else:
    unavailable.append(item)

print(available)
print(unavailable)

```

```python
# Problem	12:	Reverse	Country–City	Mapping
# Difficulty:	Hard
# Problem	Statement
# Reverse	country	→	cities	mapping	so	each	city	points	back	to	its	country.
# Input	Format
# {'Pakistan': ['Lahore', 'Karachi'], 'Turkey': ['Istanbul']}
# Output	Format
# {'Lahore': 'Pakistan', 'Karachi': 'Pakistan', 'Istanbul': 'Turkey'}
# Explanation
# Each	city	becomes	a	key	whose	value	is	the	country	name.
# Hint
# Use	nested	loops.

variable = {'Pakistan': {'Lahore', 'Karachi'},
            'Turkey': {'Istanbul'}}

new = {}

for country,cities in variable.items():
  for city in cities:
    new[city]  = country

print(new)
```

```python
# Problem 13.a : JSON-Like Path Resolver
# Difficulty: Hard

# Problem Statement
# Retrieve a value from a nested dictionary using a dot-separated path.
# Input Format
# {'user': {'address': {'city': 'Lahore'}}}
# Path: user.address.city
# Output Format
# Lahore
# Explanation
# Split the path and traverse step by step.
# Hint
# Use .split('.')

variable = {'user': {'address': {'city': 'Lahore'}}}
def path_resolver(variable):
  if type(variable) is not dict:
    return variable

  key = list(variable.keys())[0]
  value = variable.get(key)
  return path_resolver(value)

path_resolver(variable)
```

```python
# Problem	15:	Dictionary	Difference	Finder
# Difficulty:	Hard
# Problem	Statement
# Find	keys	whose	values	differ	between	two	dictionaries.
# Input	Format
# {'x': 10, 'y': 20}
# {'x': 10, 'y': 25}
# Output	Format
# {'y': (20, 25)}
# Explanation
# Only	include	mismatched	keys.
# Hint
# Compare	common	keys.

d1 = {'x': 10, 'y': 20}
d2 = {'x': 10, 'y': 25}

result = {}

for key in d1:
    if key in d2:
        if d1[key] != d2[key]:
            result[key] = (d1[key], d2[key])

print(result)
```

```python
# Problem 19: CSV User Country Analyzer
# Difficulty: Project
# Problem Statement
# Read CSV user data and count users per country.
# Input Format
# name,country
# Ali,Pakistan
# Sara,Pakistan
# John,USA
# Output Format
# {'Pakistan': 2, 'USA': 1}
# Explanation
# CSV files can be read using csv.DictReader.
# import csv
# with open('names.csv', newline='') as csvfile:
# reader = csv.DictReader(csvfile)
# for row in reader:
# print(row['name'], row['country'])
# Hint
# Use .get() to count.


import csv

country_count = {}

with open('names.csv', newline='') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        country = row['country']
        print(country)
        country_count[country] = country_count.get(country, 0) + 1

print(country_count)
```

```python
# Problem 23: Log File Error Counter
# Difficulty: Project

# Problem Statement
# Analyze a log file and count different error types.
# Input Format
# ERROR: DB failed
# WARNING: Low memory
# ERROR: Timeout
# Output Format
# {'ERROR': 2, 'WARNING': 1}
# Explanation
# Read file line by line.
# Hint
# Split lines and count prefixes.

logs = [
    "ERROR: DB failed",
    "WARNING: Low memory",
    "ERROR: Timeout"
]

error_count = {}

for line in logs:
    # Extract the prefix before ':'
    level = line.split(":")[0]

    if level in error_count:
        error_count[level] += 1
    else:
        error_count[level] = 1

print(error_count)
```

```python
# Problem 24: Sales Aggregation System
# Difficulty: Project

# Problem Statement
# Aggregate total sales per product from transaction data.
# Input Format
# [('apple',50),('banana',30),('apple',20)]
# Output Format
# {'apple': 70, 'banana': 30}
# Explanation
# Combine duplicate product entries.
# Hint
# Accumulate values.


data = [('apple',50), ('banana',30), ('apple',20)]

sales = {}

for product, amount in data:
    # print(product,amount)
    if product in sales:
        sales[product] += amount
    else:
        sales[product] = amount

print(sales)

```

```python
# Problem 25: API Response Normalizer
# Difficulty: Project

# Problem Statement
# Flatten a nested API response into dot-separated keys.
# Input Format
# {'user':{'profile':{'email':'a@x.com'}}}
# Output Format
# {'user.profile.email':'a@x.com'}
# Explanation
# Traverse nested dictionaries recursively.
# Hint
# Use recursion.

def flatten_dict(data, parent_key='', result=None):
    if result is None:
        result = {}

    for key, value in data.items():
        new_key = parent_key + '.' + key if parent_key else key

        if isinstance(value, dict):
            flatten_dict(value, new_key, result)
        else:
            result[new_key] = value

    return result


api = {'user': {'profile': {'email': 'a@x.com'}}}

output = flatten_dict(api)
print(output)
```

```python
# 18. Digit Frequency Counter
# Problem	Statement:
# Given	an	integer	number,	count	how	many	times	each	digit	(0–9)	appears	in	it	using	a	for
# loop	or	a	while	loop.
# This	problem	helps	you	practice	looping	through	numbers,	handling	digits,	and	using	lists
# for	frequency	counting.
# Rules:
# Ignore	the	negative	sign	if	the	number	is	negative.
# Display	all	digits	(0–9),	even	if	their	count	is	zero.
# You	can	use	either:
# A	for	loop	to	iterate	through	the	digits	as	a	string,	or
# A	while	loop	with	modulus	(%)	and	division	(//).
# Input	Format:
# An	integer	n.
# Output	Format:a
# Display	each	digit	from	0	to	9	followed	by	its	frequency	count.

# Examples:
# • Input:	105550
# Output:
# 0	→ 2
# 1	→ 1
# 2	→ 0
# 3	→ 0
# 4	→ 0
# 5	→ 3
# 6	→ 0
# 7	→ 0
# 8	→ 0
# 9	→ 0


n = int(input("Enter a number: "))

# n = abs(n)
# n = n*-1 if n < 0 else n
if n < 0:
    n = -n

freq = [0] * 10

if n == 0:
    freq[0] = 1
else:
    while n > 0:
        d = n % 10
        freq[d] += 1
        n //= 10

for i in range(10):
    print(f"{i} → {freq[i]}")
```

```python
# 38. Check Strong Number
# A Strong Number is a number in which the sum of factorials of its digits equals the
# number itself.
# Input Format:
# An integer n.
# Output Format:
# Print "Strong" if the number is strong, otherwise "Not Strong".
# Examples:
# • Input: 145 → Output: Strong
# (Explanation: 1! + 4! + 5! = 145)
# • Input: 123 → Output: Not Strong

n = int( input() )
t = n
s = 0

while t > 0:
    d = t % 10
    f = 1
    for i in range(1, d + 1):
        f *= i
    s += f
    t //= 10

if s == n:
    print("Strong")
else:
    print("Not Strong")



```

```python

# # 39. Check Perfect Number
# # A Perfect Number is one that is equal to the sum of its proper divisors (excluding the
# # number itself).
# # Input Format:
# # An integer n.
# # Output Format:
# # Print "Perfect" if the number is perfect, otherwise "Not Perfect".
# # Examples:

# • Input: 6 → Output: Perfect
# (Explanation: 1 + 2 + 3 = 6)
# • Input: 10 → Output: Not Perfect

n = int( input() )
s = 0

for i in range(1, n):
    if n % i == 0:
        s += i

if s == n:
    print("Perfect")
else:
    print("Not Perfect")
```

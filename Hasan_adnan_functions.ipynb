```python
# Problem 21: Loan EMI Calculator
# Problem	Statement:
# Write	a	function	that	calculates	the	monthly	EMI	(Equated	Monthly	Installment) for	a
# loan	using	the	standard	EMI	formula:
# Where:
# • P =	Loan	amount	(Principal)
# • r =	Monthly	interest	rate	(in	decimal	form,	e.g.,	1%	=	0.01)
# • n =	Number	of	monthly	installments
# You	must	not	use	Python’s	built-in	power	operator	(**)	or	pow().
# Instead,	create	your	own	power	function	and	use	it
# (1 + r)n
# After	computing	the	EMI,	print	the	result	rounded	to	1–2	decimal	places.
# Input	Format:
# Principal	P,	monthly	interest	rate	r	(in	decimal),	and	number	of	months	n.
# Output	Format:
# Print	the	EMI	rounded	to	a	suitable	number	of	decimal	places.
# Examples:
# Input:	100000	0.01	12	→	Output:	approximately	8884.0

def loan():
    P = float(input("Enter P: "))
    r = float(input("Enter r: "))
    n = int(input("Enter n: "))

    A = 1
    for _ in range(n):
        A *= (1 + r)

    EMI = (P * r * A) / (A - 1)

    print(round(EMI, 2))  #round used round the emi  2 decimal places

loan()
```

```python
# Problem 25: Advanced Temperature Metrics Analyzer
# Problem	Statement:
# Write	a	function	that	reads	temperatures	until	-999	and	prints	highest,	lowest,	count,
# average,	and	count	of	freezing	readings	(<5°C).
# Input	Format:
# Multiple	temperatures	ending	with	-999.
# Output	Format:
# Return these	metrics	in	a	clear	format.
# Examples:
# Input:	3	8	-2	15	-999	→	Return:	Highest=15	Lowest=-2	Count=4	Average=6	Freezing=2

def problem25():
    highest = None
    lowest = None
    count = 0
    total = 0
    freezing = 0
    t = -1
    while t != -999:
        t = float(input())

        # if t == -999:
        #     break

        if highest is None or t > highest:
            highest = t

        if lowest is None or t < lowest:
            lowest = t

        count += 1
        total += t

        if t < 5:
            freezing += 1

    avg = total / count if count > 0 else 0
    print(f"Highest={int(highest)} Lowest={int(lowest)} Count={count} Average={avg} Freezing={freezing}")

problem25()
```

```python
# Problem 28: Inventory Countdown Simulation
# Problem	Statement:
# Write	a	function	that	simulates	reducing	item	stock	each	day	based	on	sales	until	stock
# reaches	zero	or	below.
# Input	Format:
# Initial	stock	S	followed	by	daily	sales	quantities.
# Output	Format:
# Print	the	day	on	which	stock	finishes.
# Examples:
# Input:	S=20,	sales:	3	4	5	6	4	→	Output:	Stock	finished	on	day	4





















def problem28():
  stock = int(input("enter stock"))
  day_count = 0
  while stock > 0:
    sales = int(input("Enter daily sale: "))
    day_count += 1
    stock -= sales



  print(f"Stock finished on day {day_count-1}")


problem28()
```

```python
# Problem 29: Numeric Pyramid Generator
# Problem	Statement:
# Write	a	function	that	prints	a	centered	numeric	pyramid	of	height	n,	where	each	row	forms
# a	symmetric	number	pattern.
# Input	Format:
# An	integer	n.
# Output	Format:
# Print	the	pyramid	pattern.
# Examples:
# Input:	3	→	Output:
# 				1
# 121

def problem29(n):
    r = 1
    while r <= n:
        s = 1
        while s <= n - r:
            print(" ", end="")
            s += 1

        a = 1
        while a <= r:
            print(a, end="")
            a += 1

        b = r - 1
        while b >= 1:
            print(b, end="")
            b -= 1

        print()
        r += 1

problem29(3)

```

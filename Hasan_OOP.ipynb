```python
# Problem 9: Multilevel Inheritance Example
# Difficulty: Hard
# Problem Statement:
# Create GrandParent, Parent, and Child classes. Each class should have one method. Call all
# methods using a Child object.
# Output:
# GrandParent method
# Parent method
# Child methoda

class GrandParent:
    def gp_method(self):
        print("GrandParent method")

class Parent(GrandParent):
    def p_method(self):
        print("Parent method")

class Child(Parent):
    def c_method(self):
        print("Child method")

ch = Child()
ch.gp_method()
ch.p_method()
ch.c_method()
```

```python
# Problem 10: Shape Area Calculator
# Difficulty: Expert
# Problem Statement:
# Create a Shape base class. Create Rectangle and Circle classes that inherit Shape and
# calculate area.

# Output:
# Rectangle Area = 20
# Circle Area = 78.5

class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

r = Rectangle(5, 4)
c = Circle(5)
print("Rectangle Area =", int(r.area()))
print("Circle Area =", c.area())
```

```python
# Problem 13: Student Management System
# Difficulty: Project
# Problem Statement:
# Create a Student system that stores name, marks, and calculates grade. Display full student
# report.
# Output:
# Name: Ali, Marks: 85, Grade: A

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def grade(self):
        if self.marks >= 80:
            return "A"
        if self.marks >= 70:
            return "B"
        if self.marks >= 60:
            return "C"
        if self.marks >= 50:
            return "D"
        return "F"

    def report(self):
        print("Name:", self.name + ",", "Marks:", str(self.marks) + ",", "Grade:", self.grade())

s = Student("Ali", 85)
s.report()
```

```python
# Problem 14: Bank Account Types
# Difficulty: Project
# Problem Statement:
# Create Account base class. SavingsAccount adds interest, CurrentAccount deducts fee.

# Display final balance.
# Output:
# Final Balance: 1050


class Account:
    def __init__(self, balance):
        self.balance = balance

class SavingsAccount(Account):
    def add_interest(self):
        self.balance += self.balance * 0.05

class CurrentAccount(Account):
    def deduct_fee(self):
        self.balance -= 50

sa = SavingsAccount(1000)
sa.add_interest()
print("Final Balance:", int(sa.balance))
```

```python
# Problem 15: Vehicle Rental System
# Difficulty: Project
# Problem Statement:
# Create Vehicle base class. Bike and Car calculate rent per day differently. Calculate total rent
# based on number of days.
# Input:
# Days = 3
# Output:
# Total Rent = 3000

class VehicleRental:
    def rent_per_day(self):
        return 0

    def total_rent(self, days):
        return self.rent_per_day() * days

class BikeRental(VehicleRental):
    def rent_per_day(self):
        return 500

class CarRental(VehicleRental):
    def rent_per_day(self):
        return 1000

days = 3
cr = CarRental()
print("Total Rent =", cr.total_rent(days))
```

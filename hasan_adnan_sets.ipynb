```python
# Problem	1:	Remove	Duplicates	from	a	List
# Difficulty:	Easy
# Problem	Statement
# Given	a	list	of	integers,	remove	duplicate	values	using	a	set	and	return	the	unique	set	of
# numbers.
# Input	Format
# A	list	of	integers.
# Output	Format
# A	set	of	unique	integers.
# Example
# Input:	[1,2,2,3,1,4]
# Output:	{1,2,3,4}
# Explanation
# Sets	automatically	remove	duplicates

arr = [1,2,2,3,1,4]
arr = set(arr)
print(arr)
```

```python
# Problem	2:	Check	Membership	in	a	Set
# Difficulty:	Easy
# Problem	Statement
# Check	if	a	given	value	exists	in	a	set.
# Input	Format
# A	set	and	an	integer.
# Output	Format
# Example
# Set={3,6,9},wValue=6	→	Present
# Explanation
# Membership	check	uses	O(1)	lookup

#Method 1

my_set = {3,6,9}
value = int(input("Enter the value to check"))
if value in my_set:
  print("Present")
else:
  print("Not present")

```

```python
# Problem	3:	Count	Unique	Words
# Difficulty:	Easy
# Problem	Statement
# Count	unique	words	in	a	sentence	using	a	set.
# Input	Format
# A	sentence	string.
# Output	Format
# Integer	count.
# Example
# Input:	'this	is	a	test	this	is	good'	→	5
# Explanation
# Convert	words	to	set;	duplicates	removed

#Method 1

sentence = input("enter a string:")
word = sentence.split()
unique_words = set(word)
print(len(unique_words))



# #Method 2

# sentence = input("Enter a string: ")
# words = sentence.split()
# print(words)
# unique_set = set(words)
# print(unique_set)
# print(len(unique_set))

```

```python
# Problem	4:	Compare	Two	Sets	by	Size
# Difficulty:	Easy
# Problem	Statement
# Convert	two	lists	to	sets	and	print	the	larger	set.
# Input	Format
# Two	lists.
# Output	Format
# A	set	with	more	unique	values.
# Example
# Input:	[1,2,3,3]	&	[4,5]	→	{1,2,3}
# Explanation
# Set	with	more	unique	items	is	printed.

arr1 = set([1,2,3,3])
arr2 = set([4,5])
if len(arr1) > len(arr2):
  print(arr1)
else:
  print(arr2)
```

```python
# Problem	5:	Set	Length	Calculator
# Difficulty:	Easy
# Problem	Statement
# Print	number	of	elements	in	a	set.
# Input	Format
# A	set.
# Output	Format
# Integer	length.
# Example
# Input:	{5,10,15}	→	3
# Explanation
# len()	returns	number	of	elements

arr = set([5,10,15])
print(len(arr))
```

```python
# Problem	6:	Common	Elements
# Difficulty:	Medium
# Problem	Statement
# Find	intersection	of	two	sets.
# Input	Format
# Two	sets.
# Output	Format
# Intersection	set.
# Example
# Input:	{1,2,3,4}	&	{3,4,5}	→	{3,4}
# Explanation
# Intersection	gives	common	entries

arr1 = set([1,2,3,4])
arr2 = set([3,4,5])
common_arr = arr1.intersection(arr2) # intersection give common between them
print(common_arr)
```

```python
# Problem	7:	Unique	Elements	of	First	Set
# Difficulty:	Medium
# Problem	Statement
# Find	A	−	B.
# Input	Format
# Two	sets	A	and	B.
# Output	Format
# Difference	set.
# Example
# A={10,20,30},	B={20,40}	→	{10,30}
# Explanation
# Difference	removes	B's	elements	from	A.

A = set([10,20,30])
B = set([20,40])
difference_ab = A.difference(B) # difference is used to remove common between two sets
print(difference_ab)
```

```python
# Problem	8:	Symmetric	Difference
# Difficulty:	Medium
# Problem	Statement
# Find	elements	in	either	set	but	not	both.
# Input	Format
# Two	sets.
# Output	Format
# Symmetric	difference.
# Example
# {1,2,3}	&	{3,4}	→	{1,2,4}
# Explanation
# Elements	appearing	in	exactly	one	set


a = set([1,2,3])
b = set([3,4])


print(a.symmetric_difference(b))  # this is used to find symmetric difference
```

```python
# Problem	9:	Subset	Check
# Difficulty:	Medium
# Problem	Statement
# Check	if	A	is	subset	of	B.
# Input	Format
# Two	sets.
# Output	Format
# Subset	or	Not	Subset.
# Example
# A={1,2},	B={1,2,3}	→	Subset
# Explanation
# A	is	subset	if	all	elements of a	exist	in	B

a = set([1,2])
b = set([1,2,3])

a_b = a.issubset(b)
if a_b == True:
  print("Subset")
else:
  print("Not Subset")
```

```python
# Problem	10:	Unique	Characters
# Difficulty:	Medium
# Problem	Statement
# Extract	unique	characters	using	a	set.
# Input	Format
# String.
# Output	Format
# Set	of	characters.
# Example
# 'banana'	→	{'b','a','n'}
# Explanation
# Set	removes	repeated	letters.

a = set("banana")
print(a)
```

```python
# Problem	11:	Students	in	Both	Courses
# Difficulty:	Hard
# Problem	Statement
# Find	students	enrolled	in	both	A	and	B.
# Input	Format
# Two	sets	of	names.
# Output	Format
# Intersection.
# Example
# A={'Ali','Sara','John'},	B={'Sara','Mike','John'}	→	{'Sara','John'}
# Explanation
# Intersection	gives	common	names

A = set(['Ali','Sara','John'])
B = set(['Sara','Mike','John'])

common = A.intersection(B)
print(common)

```

```python
# Problem	12:	Missing	Days
# Difficulty:	Hard
# Problem	Statement
# Find	days	student	did	not	attend.
# Input	Format
# Set	of	attended	days.
# Output	Format
# Set	of	missing	days.
# Example
# Full={Mon,Tue,Wed,Thu,Fri},	Attended={Mon,Wed}	→	{Tue,Thu,Fri}
# Explanation
# Difference	gives	missing	elements.

Full = set(["Mon","Tue","Wed","Thu","Fri"])
attended = set(["Mon","Wed"])

missed_day = Full.symmetric_difference(attended)
print(missed_day)
```

```python
# Problem	13:	Duplicate	Words	Across	Sentences
# Difficulty:	Hard
# Problem	Statement
# Find	common	words	across	two	sentences.
# Input	Format
# Two	strings.
# Output	Format
# Set	of	shared	words.
# Example
# 'Python	is	great'	&	'Great	learning	python'	→	{'python','great'}
# Explanation
# Lowercase	+	intersection	of	word	sets.

# Hasan Solution

sentence_1 = "Python is great"
sentence_1 = sentence_1.lower()
sentence_1_splitting = sentence_1.split()
final_sentence = set(sentence_1_splitting)

sentence_2 = "Great Learning python"
sentence_2 = sentence_2.lower()
sentence_2_splitting = sentence_2.split()
final_sentence_2 = set(sentence_2_splitting)


common = final_sentence.intersection(final_sentence_2)
print(common)



# # a = set(('Python	is great').lower().split())
# sentence1 = 'Python	is great'
# print(sentence1)
# sentence1 = sentence1.lower()
# print(sentence1)
# words_list1 = sentence1.split()
# print(words_list1)
# words_set1 = set(words_list1)
# print(words_set1)
# # b = set(("Great	learning	python").lower().split())
# sentence2 = 'Great	learning	python'
# print(sentence2)
# sentence2 = sentence2.lower()
# print(sentence2)
# words_list2 = sentence2.split()
# print(words_list2)
# words_set2 = set(words_list2)
# print(words_set2)

# print(words_set1.intersection(words_set2))

```

```python
# Problem	14:	Union	of	Three	Lists
# Difficulty:	Hard
# Problem	Statement
# Find	union	of	three	lists	using	sets.
# Input	Format
# Three	lists.
# Output	Format
# Union	set.
# Example
# [1,2],[2,3],[3,4]	→	{1,2,3,4}
# Explanation
# Union	combines	all	unique	values.

a = set([1,2])
b = set([2,3])
c = set([3,4])

a_b = a.union(b)
a_c = a_b.union(c)
print(a_c)

print(a.union(b).union(c))
```

```python
# Problem	15:	Disjoint	Set	Check
# Difficulty:	Hard
# Problem	Statement
# Check	if	two	sets	share	no	elements.
# Input	Format
# Two	sets.
# Output	Format
# Disjoint	or	Not.
# Example
# {1,2}	&	{3,4}	→	Disjoint
# Explanation
# Empty	intersection	means	disjoint

# Hasan Method
set_a = set([1,2])
set_b = set([3,4])

print("Disjoint" if len(set_a.intersection(set_b)) == 0 else print("Not"))


# a = set([1,2])
# b = set([3,4])


# print("Disjoint" if len(a.intersection(b)) == 0 else "Not Disjoint")
```

```python
# Problem	16:	Unique	Email	Domains
# Difficulty:	Expert
# Problem	Statement
# Extract	unique	domains	from	email	list.
# Input	Format
# List	of	emails.
# Output	Format
# Set	of	domains.
# Example
# ['a@gmail.com','b@yahoo.com','c@gmail.com']	→	{'gmail.com','yahoo.com'}
# Explanation
# Extract	domain	part	after	'@'

# Hasan Method
emails = ['a@gmail.com','b@yahoo.com','c@gmail.com']
domains = [item.split("@")[1] for item in emails]
print(set(domains))



# emails = ['a@gmail.com','b@yahoo.com','c@gmail.com']

# domains = [item.split("@")[1] for item in emails]

# print(domains)
# print(set(domains))
```

```python
# Problem	17:	Items	Not	Bought
# Difficulty:	Expert
# Problem	Statement
# Find	items	nobody	purchased.
# Input	Format
# Full	item	set	+	list	of	purchase	sets.
# Output	Format
# Set	of	unbought	items.
# Example
# Full={A,B,C,D},	Purchases=[{A,B},{B,C}]	→	{D}
# Explanation
# Union	all	bought	items,	subtract	from	full

full = set(["A","B","C","D"])
purchase_1 = set(["A","B"])
purchase_2 = set(["B","C"])
total_purchase = purchase_1.union(purchase_2)
final = full.difference(total_purchase)
print(final)



```

```python
# Problem	20:	Users	Visiting	All	Days
# Difficulty:	Expert
# Problem	Statement
# Find	users	who	visited	every	day.
# Input	Format
# List	of	sets	per	day.
# Output	Format
# Inersection	set.
# Example
# Day1={1,2,3},	Day2={2,3},	Day3={3,4}	→	{3}

day_1 = set([1,2,3])
day_2 = set([2,3])
day_3 = set([3,4])

c = day_1.intersection(day_2)
ans = c.intersection(day_3)

print(ans)
```

```python
# Problem	19:	Plagiarism	Detector
# Difficulty:	Expert
# Problem	Statement
# Count	overlapping	words	between	two	texts.
# Input	Format
# Two	strings.
# Output	Format
# Integer	count.
# Example
# 'data	structures	in	python'	&	'python	structures	and	algorithms'	→	2
# Explanation
# Common	words	form	intersection.

text1 = "data structures in python"
text2 = "python structures and algorithms"

words1 = set(text1.split())
words2 = set(text2.split())

common_words = words1.intersection(words2)
print(len(common_words))
```
